```python
"""
=============================================================
GROUP NO. 24-25 — ML Customer Segmentation | EASTC BDS Year III
Complete Analysis Pipeline: Preprocessing → Clustering → Comparison
=============================================================
"""

# ─────────────────────────────────────────────
# IMPORTS (All libraries organized by category)
# ─────────────────────────────────────────────

# Data manipulation & numerical computing
import pandas as pd
import numpy as np

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Machine Learning: Preprocessing
from sklearn.preprocessing import MinMaxScaler

# Machine Learning: Clustering
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN

# Machine Learning: Dimensionality Reduction
from sklearn.decomposition import PCA

# Machine Learning: Metrics
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score,
    adjusted_rand_score,
    normalized_mutual_info_score
)

# Hierarchical clustering visualization
from scipy.cluster.hierarchy import dendrogram, linkage

# Utilities
import warnings
warnings.filterwarnings('ignore')

print("✅ All libraries imported successfully")
print("="*70)
```

```python
"""
=============================================================
GROUP NO. 25 — ML Customer Segmentation | EASTC BDS Year III
Phase 2: Data Preprocessing + Feature Engineering
=============================================================
"""

# ─────────────────────────────────────────────
# PHASE 2: PROGRESS TRACKING
# ─────────────────────────────────────────────
phases = ['Data Loading', 'Missing Values', 'Encoding', 'Feature Engineering', 
          'Scaling', 'Visualizations']
progress = {phase: False for phase in phases}

# ─────────────────────────────────────────────
# 1. LOAD DATA
# ─────────────────────────────────────────────
df = pd.read_excel('survey_dataset_1523-2.xlsx')

df.columns = [
    'Timestamp','Age','Gender','District','Occupation','Network',
    'Years_Used','Monthly_Spend','MoMo_Transactions','Call_Minutes_Week',
    'SMS_Week','Primary_Service','Recharge_Frequency','Changed_Network',
    'Reason_Changed','Satisfaction','Heard_AI','AI_for_Telecom','Agree_Personalized'
]

print(f"Dataset loaded: {df.shape[0]} rows × {df.shape[1]} columns")
progress['Data Loading'] = True

# ─────────────────────────────────────────────
# 2. HANDLE MISSING VALUES
# ─────────────────────────────────────────────
# MoMo_Transactions: missing = respondent doesn't use MoMo → fill with "None/0"
df['MoMo_Transactions'] = df['MoMo_Transactions'].fillna('None')

# SMS_Week: missing → fill with "None"  
df['SMS_Week'] = df['SMS_Week'].fillna('None')

# Satisfaction: 3 missing → fill with median (mode is also acceptable)
df['Satisfaction'] = df['Satisfaction'].fillna(df['Satisfaction'].median())

# Years_Used: 1 missing → fill with mode
df['Years_Used'] = df['Years_Used'].fillna(df['Years_Used'].mode()[0])

# AI_for_Telecom & Agree_Personalized: 1-2 missing → fill with mode
df['AI_for_Telecom'] = df['AI_for_Telecom'].fillna(df['AI_for_Telecom'].mode()[0])
df['Agree_Personalized'] = df['Agree_Personalized'].fillna(df['Agree_Personalized'].mode()[0])

# Reason_Changed: 576 missing because they said "No" to changing network → fill "N/A"
df['Reason_Changed'] = df['Reason_Changed'].fillna('N/A')

print(f"Missing values after cleaning: {df.isnull().sum().sum()}")
progress['Missing Values'] = True

# ─────────────────────────────────────────────
# 3. ORDINAL ENCODING (for ML clustering)
# ─────────────────────────────────────────────

# Age groups → ordinal numbers
age_order = {'Under 18': 0, '18–24': 1, '25–34': 2, '35–44': 3, '45–54': 4, '55+': 5}
df['Age_Num'] = df['Age'].map(age_order)

# Monthly spend → ordinal (proxy for ARPU)
spend_order = {
    'Less than TZS 5,000': 1,
    '5,000–10,000': 2,
    '10,001–20,000': 3,
    '20,001–50,000': 4,
    'More than 50,000': 5
}
df['Spend_Num'] = df['Monthly_Spend'].map(spend_order)

# MoMo Transactions → ordinal (Frequency of MoMo usage)
momo_order = {'None': 0, '1–10': 1, '11–30': 2, '31–60': 3, 'More than 60': 4}
df['MoMo_Num'] = df['MoMo_Transactions'].map(momo_order)

# Call Minutes per week → ordinal (Voice usage)
call_order = {
    'Less than 30': 1, '30–60': 2, '61–120': 3, '121–300': 4, 'More than 300': 5
}
df['Call_Num'] = df['Call_Minutes_Week'].map(call_order)

# SMS per week → ordinal
sms_order = {'None': 0, '1–10': 1, '11–30': 2, 'More than 30': 3}
df['SMS_Num'] = df['SMS_Week'].map(sms_order)

# Years used → ordinal (loyalty/tenure)
tenure_order = {
    'Less than 1 year': 1, '1–2 years': 2, '3–5 years': 3, 'More than 5 years': 4
}
df['Tenure_Num'] = df['Years_Used'].map(tenure_order)

# Recharge frequency → ordinal
recharge_order = {'Occasionally': 1, 'Monthly': 2, 'Weekly': 3, 'Daily': 4}
df['Recharge_Num'] = df['Recharge_Frequency'].map(recharge_order)

# ─────────────────────────────────────────────
# 4. ONE-HOT ENCODING (categorical variables)
# ─────────────────────────────────────────────
df_encoded = pd.get_dummies(df, columns=['Gender', 'District', 'Occupation',
                                          'Network', 'Primary_Service',
                                          'Changed_Network', 'Heard_AI',
                                          'AI_for_Telecom', 'Agree_Personalized'],
                             drop_first=False)

print(f"After encoding: {df_encoded.shape[1]} columns")
progress['Encoding'] = True

# ─────────────────────────────────────────────
# 5. RFM FEATURE ENGINEERING
# ─────────────────────────────────────────────
"""
RFM = Recency · Frequency · Monetary
In telecom survey context:
  R (Recency)   → Recharge_Num (how often they recharge)
  F (Frequency) → Call_Num + MoMo_Num + SMS_Num (activity frequency)
  M (Monetary)  → Spend_Num (monthly airtime/data spend)
"""

df['RFM_Recency']   = df['Recharge_Num']
df['RFM_Frequency'] = (df['Call_Num'] + df['MoMo_Num'] + df['SMS_Num']) / 3
df['RFM_Monetary']  = df['Spend_Num']

# Composite engagement score derived from PCA weights
from sklearn.preprocessing import StandardScaler

engagement_vars = ['Spend_Num', 'Call_Num', 'MoMo_Num', 'Recharge_Num', 'SMS_Num']
scaler_engage = StandardScaler()
engage_scaled = scaler_engage.fit_transform(df[engagement_vars])

pca_engage = PCA(n_components=1, random_state=42)
pca_engage.fit(engage_scaled)
engage_loadings = np.abs(pca_engage.components_[0])
engage_weights = engage_loadings / engage_loadings.sum()
engage_weight_series = pd.Series(engage_weights, index=engagement_vars)

print("Engagement weights from PCA:")
print(engage_weight_series.round(3))

# Apply PCA-derived weights to compute engagement score
df['Engagement_Score'] = (df[engagement_vars] * engage_weight_series).sum(axis=1)

# Loyalty indicator derived from PCA weights on tenure and satisfaction
loyalty_vars = ['Tenure_Num', 'Satisfaction']
scaler_loyalty = StandardScaler()
loyalty_scaled = scaler_loyalty.fit_transform(df[loyalty_vars])

pca_loyalty = PCA(n_components=1, random_state=42)
pca_loyalty.fit(loyalty_scaled)
loyalty_loadings = np.abs(pca_loyalty.components_[0])
loyalty_weights = loyalty_loadings / loyalty_loadings.sum()
loyalty_weight_series = pd.Series(loyalty_weights, index=loyalty_vars)

print("\nLoyalty weights from PCA:")
print(loyalty_weight_series.round(3))

df['Loyalty_Score'] = (df[loyalty_vars] * loyalty_weight_series).sum(axis=1)

print("\nNew features created:")
print(df[['RFM_Recency','RFM_Frequency','RFM_Monetary',
          'Engagement_Score','Loyalty_Score']].describe().round(2))
progress['Feature Engineering'] = True

# ─────────────────────────────────────────────
# 6. FINAL ML FEATURE MATRIX
# ─────────────────────────────────────────────
ml_features = [
    'Age_Num', 'Spend_Num', 'MoMo_Num', 'Call_Num', 'SMS_Num',
    'Tenure_Num', 'Recharge_Num', 'Satisfaction',
    'RFM_Recency', 'RFM_Frequency', 'RFM_Monetary',
    'Engagement_Score', 'Loyalty_Score'
]

X = df[ml_features].copy()
print(f"\nML Feature Matrix shape: {X.shape}")
print("No missing values in feature matrix:", X.isnull().sum().sum() == 0)

# ─────────────────────────────────────────────
# 7. FEATURE SCALING (Min-Max Normalization)
# ─────────────────────────────────────────────
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
X_scaled = pd.DataFrame(scaler.fit_transform(X), columns=ml_features)

print("\nScaled feature matrix (first 3 rows):")
print(X_scaled.head(3).round(3))
progress['Scaling'] = True


# 8. SAVE CLEANED DATA

df.to_csv('C:/Users/Mussa/Desktop/work/cleaned_survey_data.csv', index=False)
X_scaled.to_csv('C:/Users/Mussa/Desktop/work/ml_features_scaled.csv', index=False)
print("\n✓ Saved: cleaned_survey_data.csv")
print("✓ Saved: ml_features_scaled.csv")


# 9. DESCRIPTIVE STATS PLOTS

fig, axes = plt.subplots(2, 3, figsize=(15, 9))
fig.suptitle('Telecom Survey — Descriptive Analysis\nDar es Salaam (n=1,523)',
             fontsize=14, fontweight='bold')

# Network distribution
ax = axes[0, 0]
net_counts = df['Network'].value_counts()
colors = ['#2196F3','#4CAF50','#FF9800','#E91E63','#9C27B0','#00BCD4']
bars = ax.bar(net_counts.index, net_counts.values, color=colors[:len(net_counts)])
ax.set_title('Network Provider Distribution')
ax.set_ylabel('Number of Respondents')
ax.tick_params(axis='x', rotation=30)
for bar, val in zip(bars, net_counts.values):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 5,
            str(val), ha='center', fontsize=9)

# Age distribution
ax = axes[0, 1]
age_order_list = ['Under 18','18–24','25–34','35–44','45–54','55+']
age_counts = df['Age'].value_counts().reindex(age_order_list)
ax.bar(age_counts.index, age_counts.values, color='#42A5F5')
ax.set_title('Age Group Distribution')
ax.set_ylabel('Count')
ax.tick_params(axis='x', rotation=30)

# Primary service used
ax = axes[0, 2]
service_counts = df['Primary_Service'].value_counts()
wedges, texts, autotexts = ax.pie(
    service_counts.values,
    labels=service_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    colors=sns.color_palette('Set2', len(service_counts))
)
ax.set_title('Primary Service Used')

# Monthly spend
ax = axes[1, 0]
spend_order_list = ['Less than TZS 5,000','5,000–10,000','10,001–20,000',
                     '20,001–50,000','More than 50,000']
spend_counts = df['Monthly_Spend'].value_counts().reindex(spend_order_list)
ax.bar(range(len(spend_counts)), spend_counts.values, color='#66BB6A')
ax.set_xticks(range(len(spend_counts)))
ax.set_xticklabels(['<5k','5-10k','10-20k','20-50k','>50k'], rotation=0)
ax.set_title('Monthly Spend (TZS)')
ax.set_ylabel('Count')

# Satisfaction rating
ax = axes[1, 1]
sat_counts = df['Satisfaction'].value_counts().sort_index()
ax.bar(sat_counts.index.astype(int), sat_counts.values,
       color=['#EF5350','#FF7043','#FFA726','#66BB6A','#26C6DA'])
ax.set_title('Overall Satisfaction (1–5)')
ax.set_xlabel('Rating')
ax.set_ylabel('Count')
ax.set_xticks([1,2,3,4,5])

# Engagement Score distribution
ax = axes[1, 2]
ax.hist(df['Engagement_Score'], bins=20, color='#AB47BC', edgecolor='white')
ax.set_title('Engagement Score Distribution')
ax.set_xlabel('Score')
ax.set_ylabel('Frequency')

plt.tight_layout()
plt.show()
progress['Visualizations'] = True

# ─────────────────────────────────────────────
# 10. WORKFLOW PROGRESS SUMMARY
# ─────────────────────────────────────────────
print("\n" + "="*60)
print("PHASE 2 WORKFLOW PROGRESS")
print("="*60)
for i, (phase, status) in enumerate(progress.items(), 1):
    status_icon = "✅" if status else "⏳"
    print(f"{status_icon} [{i}/6] {phase}")
print("="*60)
print("✅ PHASE 2 COMPLETE — Ready for Clustering (Phase 3)")
print("="*60)
```

```python
"""
=============================================================
GROUP NO. 24 — ML Customer Segmentation | EASTC BDS Year III
SCRIPT 1 OF 3: K-MEANS CLUSTERING
=============================================================
Input : cleaned_survey_data.csv  (1,523 respondents, 13 ML features)
Output: KMeans_clustered_data.csv, KMeans_cluster_profiles.csv,
        KMeans_Elbow_PCA.png
=============================================================
"""

# ── 1. Load data & build the 13-feature ML matrix ──
df = pd.read_csv("cleaned_survey_data.csv")

FEATURES = ['Age_Num', 'Spend_Num', 'MoMo_Num', 'Call_Num', 'SMS_Num',
            'Tenure_Num', 'Recharge_Num', 'Satisfaction', 'RFM_Recency',
            'RFM_Frequency', 'RFM_Monetary', 'Engagement_Score', 'Loyalty_Score']

X = MinMaxScaler().fit_transform(df[FEATURES])
print(f"Feature matrix: {X.shape[0]} respondents x {X.shape[1]} features")

# ── 2. Elbow Method + Silhouette to select k (2-10) ──
inertia, sil_scores = [], []
for k in range(2, 11):
    km = KMeans(n_clusters=k, random_state=42, n_init=10, max_iter=300).fit(X)
    inertia.append(km.inertia_)
    sil_scores.append(silhouette_score(X, km.labels_))
    print(f"  k={k:2d} | Inertia={km.inertia_:8.2f} | Silhouette={sil_scores[-1]:.4f}")

# ── 3. Fit final model (k=4) ──
BEST_K = 4
kmeans = KMeans(n_clusters=BEST_K, random_state=42, n_init=20, max_iter=500)
df['KMeans_Cluster'] = kmeans.fit_predict(X)

sil = silhouette_score(X, df['KMeans_Cluster'])
dbi = davies_bouldin_score(X, df['KMeans_Cluster'])
ch = calinski_harabasz_score(X, df['KMeans_Cluster'])
print(f"\nK-Means (k=4) -> Silhouette={sil:.4f} | Davies-Bouldin={dbi:.4f} | Calinski-Harabasz={ch:.1f}")

# ── 4. Segment labels & profiling ──
SEGMENT_NAMES = {
    0: 'High-Spend Infrequent Rechargers',
    1: 'Low-Spend Daily Rechargers',
    2: 'Budget Passive Users',
    3: 'High-Value Heavy Users'
}
df['Segment_Label'] = df['KMeans_Cluster'].map(SEGMENT_NAMES)

profile = df.groupby('KMeans_Cluster')[FEATURES].mean().round(3)
profile.insert(0, 'N', df['KMeans_Cluster'].value_counts().sort_index())
profile.insert(1, 'Pct', (profile['N'] / len(df) * 100).round(1))
print("\nCluster profiles:\n", profile)

# ── 5. Plot: Elbow curve + PCA scatter ──
X_pca = PCA(n_components=2, random_state=42).fit_transform(X)
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Left plot: Silhouette Scores
axes[0].plot(range(2, 11), sil_scores, 'go-')
axes[0].axvline(BEST_K, color='red', linestyle='--', label=f'k={BEST_K}')
axes[0].set(title='Silhouette Score vs k', xlabel='k', ylabel='Silhouette Score')
axes[0].legend()

# Right plot: PCA Clusters
for c in range(BEST_K):
    mask = df['KMeans_Cluster'] == c
    axes[1].scatter(X_pca[mask, 0], X_pca[mask, 1], s=12, alpha=0.6, label=f'C{c}')
axes[1].set(title='K-Means Clusters (PCA)', xlabel='PC1', ylabel='PC2')
axes[1].legend(fontsize=8)

plt.tight_layout()
plt.show()
```

```python
"""
=============================================================
GROUP NO. 24 — ML Customer Segmentation | EASTC BDS Year III
SCRIPT 2 OF 3: HIERARCHICAL CLUSTERING (Ward's Linkage)
=============================================================
Input : cleaned_survey_data.csv  (1,523 respondents, 13 ML features)
Output: Hierarchical_clustered_data.csv, Hierarchical_Dendrogram.png
=============================================================
"""

# ── 1. Load data & build the 13-feature ML matrix ──
df = pd.read_csv("cleaned_survey_data.csv")

FEATURES = ['Age_Num', 'Spend_Num', 'MoMo_Num', 'Call_Num', 'SMS_Num',
            'Tenure_Num', 'Recharge_Num', 'Satisfaction', 'RFM_Recency',
            'RFM_Frequency', 'RFM_Monetary', 'Engagement_Score', 'Loyalty_Score']

X = MinMaxScaler().fit_transform(df[FEATURES])

# ── 2. Fit Agglomerative Clustering (Ward linkage, k=4) ──
BEST_K = 4
hc = AgglomerativeClustering(n_clusters=BEST_K, linkage='ward')
df['Hierarchical_Cluster'] = hc.fit_predict(X)

sil = silhouette_score(X, df['Hierarchical_Cluster'])
dbi = davies_bouldin_score(X, df['Hierarchical_Cluster'])
ch = calinski_harabasz_score(X, df['Hierarchical_Cluster'])
print(f"Hierarchical (Ward, k=4) -> Silhouette={sil:.4f} | Davies-Bouldin={dbi:.4f} | Calinski-Harabasz={ch:.1f}")
print(df['Hierarchical_Cluster'].value_counts(normalize=True).sort_index().round(3))

# ── 3. Dendrogram (on a random sample of 200, for readability) ──
sample = pd.DataFrame(X).sample(200, random_state=42).values
Z = linkage(sample, method='ward')

plt.figure(figsize=(10, 5))
dendrogram(Z, truncate_mode='lastp', p=30, leaf_rotation=90)
plt.axhline(y=Z[-BEST_K, 2], color='red', linestyle='--', label=f'Cut at k={BEST_K}')
plt.title("Hierarchical Clustering Dendrogram (Ward's Linkage, n=200 sample)")
plt.xlabel("Respondents (truncated)")
plt.ylabel("Distance")
plt.legend()
plt.tight_layout()
plt.show()

```

```python
"""
=============================================================
GROUP NO. 24 — ML Customer Segmentation | EASTC BDS Year III
SCRIPT 2B OF 3: DBSCAN CLUSTERING (Density-Based)
=============================================================
Input : cleaned_survey_data.csv  (1,523 respondents, 13 ML features)
Output: DBSCAN_clustered_data.csv, DBSCAN_visualization.png
=============================================================
"""

# ── 1. Load data & build the 13-feature ML matrix ──
df = pd.read_csv("cleaned_survey_data.csv")

FEATURES = ['Age_Num', 'Spend_Num', 'MoMo_Num', 'Call_Num', 'SMS_Num',
            'Tenure_Num', 'Recharge_Num', 'Satisfaction', 'RFM_Recency',
            'RFM_Frequency', 'RFM_Monetary', 'Engagement_Score', 'Loyalty_Score']

X = MinMaxScaler().fit_transform(df[FEATURES])

# ── 2. Tune DBSCAN (eps and min_samples) ──
# Find optimal eps using k-distance graph
from sklearn.neighbors import NearestNeighbors

neighbors = NearestNeighbors(n_neighbors=5)
neighbors_fit = neighbors.fit(X)
distances, indices = neighbors_fit.kneighbors(X)
distances = np.sort(distances[:, -1], axis=0)

# Use the elbow method: eps at 90th percentile of k-distance
eps_optimal = np.percentile(distances, 90)
min_samples = 5

print(f"DBSCAN Parameters: eps={eps_optimal:.4f}, min_samples={min_samples}")

# ── 3. Fit DBSCAN ──
dbscan = DBSCAN(eps=eps_optimal, min_samples=min_samples)
df['DBSCAN_Cluster'] = dbscan.fit_predict(X)

n_clusters_dbscan = len(set(df['DBSCAN_Cluster'])) - (1 if -1 in df['DBSCAN_Cluster'] else 0)
n_noise = list(df['DBSCAN_Cluster']).count(-1)

print(f"\nDBSCAN Results:")
print(f"  • Number of clusters: {n_clusters_dbscan}")
print(f"  • Number of noise points (outliers): {n_noise} ({n_noise/len(df)*100:.1f}%)")
print(f"  • Cluster distribution:\n{df['DBSCAN_Cluster'].value_counts().sort_index()}")

# ── 4. Compute DBSCAN Metrics (only if not all noise) ──
if n_clusters_dbscan > 1:
    dbscan_sil = silhouette_score(X, df['DBSCAN_Cluster'])
    dbscan_dbi = davies_bouldin_score(X, df['DBSCAN_Cluster'])
    dbscan_ch = calinski_harabasz_score(X, df['DBSCAN_Cluster'])
    print(f"\nDBSCAN Metrics -> Silhouette={dbscan_sil:.4f} | Davies-Bouldin={dbscan_dbi:.4f} | Calinski-Harabasz={dbscan_ch:.1f}")
else:
    print("\n⚠️  DBSCAN found too few clusters for reliable metrics. Consider adjusting eps.")
    dbscan_sil = np.nan
    dbscan_dbi = np.nan
    dbscan_ch = np.nan

# ── 5. Visualization: DBSCAN Clusters (PCA) ──
X_pca = PCA(n_components=2, random_state=42).fit_transform(X)

plt.figure(figsize=(12, 6))

# Plot clusters
unique_clusters = sorted(set(df['DBSCAN_Cluster']))
colors_dbscan = plt.cm.Spectral(np.linspace(0, 1, len(unique_clusters)))

for idx, cluster_id in enumerate(unique_clusters):
    if cluster_id == -1:
        # Plot noise points in black
        mask = df['DBSCAN_Cluster'] == cluster_id
        plt.scatter(X_pca[mask, 0], X_pca[mask, 1], s=50, c='black', 
                   marker='x', label='Noise', linewidths=2, alpha=0.7)
    else:
        mask = df['DBSCAN_Cluster'] == cluster_id
        plt.scatter(X_pca[mask, 0], X_pca[mask, 1], s=30, c=[colors_dbscan[idx]], 
                   label=f'Cluster {cluster_id}', alpha=0.6)

plt.xlabel('PC1')
plt.ylabel('PC2')
plt.title(f'DBSCAN Clustering (PCA) — {n_clusters_dbscan} clusters + {n_noise} noise points')
plt.legend(loc='best')
plt.tight_layout()
plt.show()
```

```python
"""
=============================================================
GROUP NO. 24 — ML Customer Segmentation | EASTC BDS Year III
SCRIPT 3 OF 3: ALGORITHM COMPARISON & EVALUATION (K-Means, Hierarchical, DBSCAN)
=============================================================
Input : cleaned_survey_data.csv
Output: Comprehensive metrics, visualizations, and recommendations
=============================================================
"""

# ─────────────────────────────────────────────
# 1. LOAD DATA & CLUSTER ALL THREE ALGORITHMS
# ─────────────────────────────────────────────
df = pd.read_csv("cleaned_survey_data.csv")

FEATURES = ['Age_Num', 'Spend_Num', 'MoMo_Num', 'Call_Num', 'SMS_Num',
            'Tenure_Num', 'Recharge_Num', 'Satisfaction', 'RFM_Recency',
            'RFM_Frequency', 'RFM_Monetary', 'Engagement_Score', 'Loyalty_Score']

X = MinMaxScaler().fit_transform(df[FEATURES])

# K-Means
kmeans = KMeans(n_clusters=4, random_state=42, n_init=20, max_iter=500)
kmeans_labels = kmeans.fit_predict(X)

# Hierarchical
hc = AgglomerativeClustering(n_clusters=4, linkage='ward')
hc_labels = hc.fit_predict(X)

# DBSCAN with optimized parameters
from sklearn.neighbors import NearestNeighbors
neighbors = NearestNeighbors(n_neighbors=5)
neighbors_fit = neighbors.fit(X)
distances, indices = neighbors_fit.kneighbors(X)
distances = np.sort(distances[:, -1], axis=0)
eps_optimal = np.percentile(distances, 90)
dbscan = DBSCAN(eps=eps_optimal, min_samples=5)
dbscan_labels = dbscan.fit_predict(X)

# ─────────────────────────────────────────────
# 2. COMPUTE METRICS FOR ALL THREE
# ─────────────────────────────────────────────
metrics = {}

# K-Means metrics
kmeans_sil = silhouette_score(X, kmeans_labels)
kmeans_dbi = davies_bouldin_score(X, kmeans_labels)
kmeans_ch = calinski_harabasz_score(X, kmeans_labels)
metrics['K-Means'] = {'Silhouette': kmeans_sil, 'Davies-Bouldin': kmeans_dbi, 'Calinski-Harabasz': kmeans_ch}

# Hierarchical metrics
hc_sil = silhouette_score(X, hc_labels)
hc_dbi = davies_bouldin_score(X, hc_labels)
hc_ch = calinski_harabasz_score(X, hc_labels)
metrics['Hierarchical'] = {'Silhouette': hc_sil, 'Davies-Bouldin': hc_dbi, 'Calinski-Harabasz': hc_ch}

# DBSCAN metrics
n_clusters_dbscan = len(set(dbscan_labels)) - (1 if -1 in dbscan_labels else 0)
n_noise = list(dbscan_labels).count(-1)
if n_clusters_dbscan > 1:
    dbscan_sil = silhouette_score(X, dbscan_labels)
    dbscan_dbi = davies_bouldin_score(X, dbscan_labels)
    dbscan_ch = calinski_harabasz_score(X, dbscan_labels)
else:
    dbscan_sil = dbscan_dbi = dbscan_ch = np.nan
metrics['DBSCAN'] = {'Silhouette': dbscan_sil, 'Davies-Bouldin': dbscan_dbi, 'Calinski-Harabasz': dbscan_ch}

# ─────────────────────────────────────────────
# 3. DISPLAY COMPARISON TABLE
# ─────────────────────────────────────────────
print("="*80)
print("ALGORITHM PERFORMANCE METRICS COMPARISON")
print("="*80)
comparison_df = pd.DataFrame(metrics).T
print(comparison_df.round(4))
print("\nMetric Interpretations:")
print("  • Silhouette (higher is better): Range [-1, 1], >0.5 = good separation")
print("  • Davies-Bouldin (lower is better): Avg similarity ratio, <1.5 = good")
print("  • Calinski-Harabasz (higher is better): Ratio of between/within variance")
print("="*80)

# ─────────────────────────────────────────────
# 4. CLUSTER SIZE DISTRIBUTION
# ─────────────────────────────────────────────
print("\nCLUSTER SIZE DISTRIBUTION:")
print("\n  K-Means:")
print(pd.Series(kmeans_labels).value_counts().sort_index())
print("\n  Hierarchical:")
print(pd.Series(hc_labels).value_counts().sort_index())
print(f"\n  DBSCAN ({n_clusters_dbscan} clusters + {n_noise} noise points):")
print(pd.Series(dbscan_labels).value_counts().sort_index())

# ─────────────────────────────────────────────
# 5. ALGORITHM AGREEMENT ANALYSIS
# ─────────────────────────────────────────────
ari_km_hc = adjusted_rand_score(kmeans_labels, hc_labels)
nmi_km_hc = normalized_mutual_info_score(kmeans_labels, hc_labels)
ari_km_dbscan = adjusted_rand_score(kmeans_labels, dbscan_labels)
nmi_km_dbscan = normalized_mutual_info_score(kmeans_labels, dbscan_labels)
ari_hc_dbscan = adjusted_rand_score(hc_labels, dbscan_labels)
nmi_hc_dbscan = normalized_mutual_info_score(hc_labels, dbscan_labels)

print("\n" + "="*80)
print("ALGORITHM AGREEMENT (Comparing Partitions):")
print("="*80)
print(f"  K-Means vs Hierarchical:  ARI={ari_km_hc:.4f} | NMI={nmi_km_hc:.4f}")
print(f"  K-Means vs DBSCAN:        ARI={ari_km_dbscan:.4f} | NMI={nmi_km_dbscan:.4f}")
print(f"  Hierarchical vs DBSCAN:   ARI={ari_hc_dbscan:.4f} | NMI={nmi_hc_dbscan:.4f}")
print("="*80)

# ─────────────────────────────────────────────
# 6. VISUALIZATION: METRICS COMPARISON
# ─────────────────────────────────────────────
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
fig.suptitle('Clustering Algorithm Comparison: K-Means vs. Hierarchical vs. DBSCAN', 
             fontsize=14, fontweight='bold')

# Silhouette Score
ax = axes[0, 0]
silhouette_data = [kmeans_sil, hc_sil, dbscan_sil]
colors = ['#42A5F5', '#66BB6A', '#FFA726']
bars = ax.bar(['K-Means', 'Hierarchical', 'DBSCAN'], silhouette_data, color=colors, alpha=0.7, edgecolor='black', linewidth=1.5)
ax.axhline(y=0.5, color='red', linestyle='--', label='Good threshold (0.5)', linewidth=2)
ax.set_ylabel('Silhouette Score', fontweight='bold')
ax.set_title('Silhouette Score (Higher is Better)', fontweight='bold')
ax.set_ylim([-0.2, 1])
ax.legend()
for bar, val in zip(bars, silhouette_data):
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2, height + 0.02, f'{val:.4f}', ha='center', fontweight='bold', fontsize=9)

# Davies-Bouldin Index
ax = axes[0, 1]
dbi_data = [kmeans_dbi, hc_dbi, dbscan_dbi]
bars = ax.bar(['K-Means', 'Hierarchical', 'DBSCAN'], dbi_data, color=colors, alpha=0.7, edgecolor='black', linewidth=1.5)
ax.axhline(y=1.5, color='red', linestyle='--', label='Good threshold (1.5)', linewidth=2)
ax.set_ylabel('Davies-Bouldin Index', fontweight='bold')
ax.set_title('Davies-Bouldin Index (Lower is Better)', fontweight='bold')
ax.legend()
for bar, val in zip(bars, dbi_data):
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2, height + 0.02, f'{val:.4f}', ha='center', fontweight='bold', fontsize=9)

# Calinski-Harabasz Score
ax = axes[1, 0]
ch_data = [kmeans_ch, hc_ch, dbscan_ch]
bars = ax.bar(['K-Means', 'Hierarchical', 'DBSCAN'], ch_data, color=colors, alpha=0.7, edgecolor='black', linewidth=1.5)
ax.set_ylabel('Calinski-Harabasz Score', fontweight='bold')
ax.set_title('Calinski-Harabasz Score (Higher is Better)', fontweight='bold')
for bar, val in zip(bars, ch_data):
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2, height + 10, f'{val:.1f}', ha='center', fontweight='bold', fontsize=9)

# Algorithm Agreement Matrix
ax = axes[1, 1]
agreement_matrix = np.array([[1.0, nmi_km_hc, nmi_km_dbscan],
                             [nmi_km_hc, 1.0, nmi_hc_dbscan],
                             [nmi_km_dbscan, nmi_hc_dbscan, 1.0]])
im = ax.imshow(agreement_matrix, cmap='RdYlGn', vmin=0, vmax=1, aspect='auto')
ax.set_xticks([0, 1, 2])
ax.set_yticks([0, 1, 2])
ax.set_xticklabels(['K-Means', 'Hierarchical', 'DBSCAN'], rotation=45, ha='right')
ax.set_yticklabels(['K-Means', 'Hierarchical', 'DBSCAN'])
ax.set_title('Algorithm Agreement (NMI)', fontweight='bold')

# Add text annotations
for i in range(3):
    for j in range(3):
        text = ax.text(j, i, f'{agreement_matrix[i, j]:.3f}', ha='center', va='center', 
                      color='black' if agreement_matrix[i, j] > 0.5 else 'white', fontweight='bold')

plt.colorbar(im, ax=ax, label='Normalized Mutual Info')

plt.tight_layout()
plt.show()

# ─────────────────────────────────────────────
# 7. FINAL RECOMMENDATION
# ─────────────────────────────────────────────
print("\n" + "="*80)
print("FINAL RECOMMENDATION & CONCLUSION")
print("="*80)

# Determine best algorithm (handling NaN values)
valid_scores = []
algorithms = []

if not np.isnan(kmeans_sil):
    valid_scores.append(kmeans_sil)
    algorithms.append(("K-Means", kmeans_sil))
if not np.isnan(hc_sil):
    valid_scores.append(hc_sil)
    algorithms.append(("Hierarchical", hc_sil))
if not np.isnan(dbscan_sil):
    valid_scores.append(dbscan_sil)
    algorithms.append(("DBSCAN", dbscan_sil))

best_algo, best_score = max(algorithms, key=lambda x: x[1])

print(f"\n✅ RECOMMENDED ALGORITHM: {best_algo}")
print(f"   Silhouette Score: {best_score:.4f} (highest among all algorithms)")

print(f"\n📊 ALGORITHM CHARACTERISTICS & USE CASES:")
print(f"\n   K-Means (Centroid-Based):")
print(f"   • Speed: Fast (O(nkl) - linear scaling)")
print(f"   • Shape: Assumes spherical clusters")
print(f"   • Initialization: Sensitive to starting centroids")
print(f"   • Best for: Large datasets, predefined cluster count, balanced clusters")
print(f"   • Score: Silhouette={kmeans_sil:.4f} | Davies-Bouldin={kmeans_dbi:.4f}")

print(f"\n   Hierarchical (Agglomerative, Ward):")
print(f"   • Speed: Slower (O(n²) - quadratic complexity)")
print(f"   • Shape: Can capture nested structures")
print(f"   • Output: Dendrogram for exploratory analysis")
print(f"   • Best for: Small-medium datasets, hierarchical exploration")
print(f"   • Score: Silhouette={hc_sil:.4f} | Davies-Bouldin={hc_dbi:.4f}")

print(f"\n   DBSCAN (Density-Based):")
print(f"   • Speed: Medium (O(n²) with spatial indexing)")
print(f"   • Shape: Discovers arbitrary-shaped clusters")
print(f"   • Outliers: Identifies noise/anomalies explicitly")
print(f"   • Best for: Non-spherical clusters, outlier detection")
print(f"   • Detected: {n_clusters_dbscan} clusters + {n_noise} noise points ({n_noise/len(X)*100:.1f}%)")
if not np.isnan(dbscan_sil):
    print(f"   • Score: Silhouette={dbscan_sil:.4f} | Davies-Bouldin={dbscan_dbi:.4f}")

print(f"\n💡 BUSINESS RECOMMENDATION:")
print(f"   1. Use {best_algo} for primary segmentation (k=4 segments)")
print(f"   2. Use Hierarchical for exploring segment relationships")
print(f"   3. Use DBSCAN to identify high-value outlier customers")
print(f"   4. Agreement across algorithms: {nmi_km_hc*100:.1f}% (K-Means vs Hierarchical)")
print(f"   5. Different perspective: DBSCAN identifies {n_noise} outlier customers")
print("="*80)
```
